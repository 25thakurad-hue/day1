```python
#accept 5 paper marks, rollno, name; and calculate total and percentage if percentage >=85, student can't take addmission
#using class and object and constructor
class Calculate:
    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks

    def calculate(self):
        total = sum(self.marks)
        percentage = total / 5

        print("Roll No:", self.roll_no)
        print("Name:", self.name)
        print("Total Marks:", total)
        print("Percentage:", percentage)

        if percentage >= 85:
            print("Admission not allowed")
        else:
            print("Admission allowed")


roll_no = int(input("Enter roll number: "))
name = input("Enter name: ")

marks = []
for i in range(5):
    m = float(input("Enter marks: "))
    marks.append(m)

s = Calculate(roll_no, name, marks)
s.calculate()
```
